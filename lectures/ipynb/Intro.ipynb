In Python, especially within the context of Machine Learning (ML), the None type and the del keyword have distinct and important uses.

Uses for None Type in Python ML
The None type in Python represents the absence of a value or a null value. It's often used in ML for:

- Default/Optional Arguments: Functions or class methods might use None as the default value for parameters that are optional or where the specific value should be determined later or automatically.


- Sentinel Value/Placeholder: None can act as a sentinel value to indicate that a variable hasn't been initialized, a computation hasn't been performed, or a required object is currently missing.

- Returning No Result: Functions that don't produce a meaningful value or whose main purpose is to perform an action (a side effect) often implicitly or explicitly return None.

Uses for the del Keyword in Python ML
- The del keyword is a statement used to delete names (variables) from the local or global scope and to remove items from lists or dictionaries. In ML, it's primarily used for memory management:

- Freeing Up Memory from Large Objects: In ML, you often deal with very large objects like datasets (e.g., Pandas DataFrames, NumPy arrays) or trained models. Using del explicitly removes the variable's name binding to that object. If there are no other references to the object, the Python garbage collector can then free the memory the object occupied.

- Crucial in Resource-Constrained Environments: This is vital when working with large datasets, especially on systems with limited RAM (e.g., in Google Colab or Kaggle notebooks), where you might need to load one massive file, process it, and then delete it before loading the next.


# Writting Functions

Functions are one of the most important building blocks in Python.


What is a Function?

A function is a block of reusable code that performs a specific task. Instead of writing the same logic again and again, we can “package” it into a function and call it whenever needed. They usually take input(s) (parameters) and usually produce some output.

## Why Write Functions?

### 1. Three R's, Readabilty, Reuseabilty, and avoid Repetition.

#### Activity 4: Please print out all the numbers from 1 to 5 squared below!

```python
print("---Without a function---")
print(1 * 1)
print(2 * 2)
print(3 * 3)
print(4 * 4)
print(5 * 5)

print("---With a function---")
def square(num):
    return num * num

print(square(1))
print(square(2))
print(square(3))
print(square(4))
print(square(5))
```

> Aside: In python, you can also sqaure numbers with `x**2` !

We always want to write the cleanest code possible, which often means reducing redundancy, for example, having too many print statements. So how can we simplify this? By using loops!

Loops allow us to repeatedly execute a block of code until a certain condition is met.

- A `for` loop repeats a set number of times, usually based on a range or collection.

- A `while` loop continues running until a condition becomes false (and can even run indefinitely if that condition is never met).

By using loops, we make our code shorter, easier to read, and more efficient.

```python
#note range is how we set our start and end values.
#range(start, end). So we start at 1 then go 2,3,4,5 and then when we reach 6 we stop
for i in range(1,6):
    print(square(i))
```

While our function is simple here, as you might expect, more complex code often involves repeated actions, such as calculating certain formulas for different pieces of data. These repeated actions can be simplified using functions. You will encounter many examples like this throughout the course and will also have the opportunity to write your own!

 ### 2. Organize Complex Tasks

#### Activity 5: Break down big problems into smaller, manageable pieces. Here, lets try to find the area and parameter of two rectangles.

```python
def get_area(length, width):
    return length * width

def get_perimeter(length, width):
    return 2 * (length + width)

print("---Rectangle 1---")
print("Area:", get_area(5, 3))
print("Perimeter:",get_perimeter(5, 3))

print("---Rectangle 2---")
print("Area:",get_area(50, 40))
print("Perimeter:",get_perimeter(50, 40))
```

Key Takeaway

- Coordinates:

    - The X (Easting) and Y (Northing) axes (labeled with coordinates like 30668 to 30671 for X and 48 to 52 for Y) represent the horizontal position on the ground (likely in meters, often using a projected coordinate system like UTM).

    - The Z (Elevation) axis (ranging from about 17.5 to 20.5, possibly in meters) represents the vertical height or elevation of the points above a reference datum.

    - Color-Coding: The points are colored based on their Z-value (elevation), with dark blue representing lower elevations (likely the ground or understory) and red representing higher elevations (the tops of the tree canopy).

Essentially, it's a 3D map of the physical structure of the forest, showing the ground, the trees, and the canopy.

Several insights and conclusions can be drawn from examining this LiDAR point cloud:

1. Forest Structure and Density
- Canopy Height: The vertical extent of the point cloud (from the lowest dark blue points to the highest red points) indicates the height of the trees. The maximum elevation is around ≈20.5 meters, suggesting a relatively tall canopy in this area.

- Canopy Irregularity: The top surface is highly irregular and rugged, which is typical of a natural forest. The peaks (red/orange) represent the highest points of individual or clustered tree crowns.

- Tree Density and Gaps: Areas with a dense concentration of points high up (orange/red) indicate a dense canopy. Gaps or lower sections within the canopy surface suggest openings in the forest or shorter, younger trees.

2. Terrain and Topography
- Ground Slope: By looking at the pattern of the lowest (dark blue) points, one can assess the topography of the terrain. If the lowest points are roughly flat across the X-Y plane, the ground is flat. If the minimum Z-values vary significantly, it indicates slopes.
