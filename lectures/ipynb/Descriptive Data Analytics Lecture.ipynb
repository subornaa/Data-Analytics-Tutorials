For example, calculating the mean of a column in a dataset. In other words, aggregating data.

Luckily for us, Pandas has the ability to calculate these statistics already built in.

It is worth noting that unless it is told not to, pandas will skip NaN values, excluding them from calculations. However, in this chapter we will be relying on a dataset that we have already cleaned.

The bars will be grouped by location and hue/color will be the variables  (PM10, CO, NO, NO2, and O3). However, most visualization libraries including Seaborn can only handle a few variables (no more than three).

Our data is currently stored in 'wide-form' format, which spreads variables over several columns. We need to convert our data to 'long-form' format. For the DataFrame with mean values by measurement and location, the column names could look like 'variable', 'location', and 'mean_value' after conversion.

Let's see how this works... (Within the code, "dataMean" was used as the DataFrame name to hold the calculated mean values. You should use your chosen name accordingly.)

```python
# Wide format
dataMean = data
dataMean.head()
```

```python
# Long format
melted = dataMean.melt(id_vars='location', value_vars=['PM10', 'CO', 'NO', 'NO2', 'O3'],
                       var_name='measurement', value_name='mean_value')
melted.head()
```

The method `.melt()` unpivots specified columns. As a result we end up having five records in our new DataFrame per location, instead of one.

To help you understand the method: <br />

<em>`id_vars`: Specifiy the column(s) kept as identifier <br />
`value_vars`: The columns to unpivot <br />
`var_name`: Name for the new variable column, we called it 'measurement' <br />
`value_name`: Name for the new value column, we called it 'mean_value' <br /></em>

Let's show you why we did this... We can now plot a grouped barplot, like in the example from the plot gallery!

```python
# Source: https://seaborn.pydata.org/examples/grouped_barplot.html

plt.figure(figsize = (15,8))

ax = sns.catplot(
    data=melted, kind="bar",
    x="location", y="mean_value", hue="measurement",
    errorbar="sd", palette="dark", alpha=.6, height=6)

# In case of any errors use this code:
#ax = sns.catplot(
    #data=melted, kind="bar",
    #x="location", y="mean_value", hue="measurement",
    #ci="sd", palette="dark", alpha=.6, height=6)

plt.title('Mean measurements', fontsize=20)
plt.xticks(rotation=45) # Rotates the x-axis labels
```

## Unique Values, Value Counts and Membership

Unique values, value counts and membership refers to a type of summary statistics that offers us information about the objects in a dataset, instead of the values of those objects. To illustrate this consider the following example.

```python
#using .unique to look at 'Vancouver International Airport #2' NO2 values
dataSelect = data.loc[data['location']=='Vancouver International Airport #2']
dataSelect['NO2']....()
```

In this example, the pandas function `.unique()` is used on the NO2 column, and returns all the unique values in that column. That is, duplicates are removed to give us an understanding of the spread of the data.

Therefore, these types of functions help us understand the data points within a dataset without using their values to give us further information.

Now, let's use `.value_counts()` to see how many times each of the above values appears in this column.

```python
#using .value_counts to look at 'Vancouver International Airport #2' NO2 values

dataSelect['NO2']....()
```

The final method in this category is `.isin()`, which is a boolean function similar to `.isnull()` but we can query any list containing a value or values that we are interested in. This can be very useful for filtering datasets. For example, lets say we want to look for all records with NO2 values in the following list: [40, 41, 42, 43, 44, 45, 46, 47, 48].


```python
#using .isin to look at NO2 values over 40 as bool

dataSelect['NO2']....([40, 41, 42, 43, 44, 45, 46, 47, 48])
```

 After that, we are able to use the series with boolean values to select the rows in the 'dataSelect' DataFrame.

```python
#showing the objects over 40 and their location

dataSelect.loc[dataSelect['NO2']....([40, 41, 42, 43, 44, 45, 46, 47, 48]), :]
```

## Air Polution by Day of the Week

Let's assume some of the air pollution is to do with industry in the area. In that case, pollution is likely lower during the weekend. We can use the 'date' variable in the DataFrame to work out what day of the week a record was recorded. Afterwards, we can investigate the assumption using data aggregation and visualization.



```python
#creating a datetime column with .to_datetime

data['datetime'] = pd.to_datetime(data['Date'])
data
```

Now, we can use the datetime column we just created and the function `.dayofweek` to create a column of the day of the week. Note: 0 is Monday, and 6 is Sunday.

```python
#creating a day of the week column with .dayofweek from datetime column
data['datetime'] = pd.to_datetime(data['datetime'])
data['day_of_week'] = data['datetime'].dt....
data
```

Finally, we can use our new day of the week column and `.groupby` to reorganize our data. Then we can calculate mean values by day of the week.

```python
#data.groupby('day_of_week').mean()

# In case of any errors use this code:
data.drop(columns=['Date', 'Time', 'location'])....('...').mean()
```

Now, we can also visualize this information. Let's create a graph for each of the variables with boxplots for each day of the week.

```python
# Source: https://seaborn.pydata.org/examples/grouped_barplot.html

# Plot boxplots for all four variables.
# Solution 2: plotting using a grid
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15,8))

# ax 0
sns.boxplot(ax=axes[0,0], data=data, x='day_of_week', y='PM10', whis=1.5)


# # ax 1
sns.boxplot(ax=axes[0,1], data=data, x='day_of_week', y='CO', whis=1.5)


# # ax 2
sns.boxplot(ax=axes[1,0], data=data, x='day_of_week', y='NO', whis=1.5)


# # ax 3
sns.boxplot(ax=axes[1,1], data=data, x='day_of_week', y='NO2', whis=1.5)


# # ax 4
sns.boxplot(ax=axes[2,1], data=data, x='day_of_week', y='O3', whis=1.5)

plt.tight_layout()
```

What are some trends we can observe from these plots?

*Answer here*

The graphs suggest that traffic-related pollutants like NO are marginally lower on the weekend, while Ozone (O3) concentrations are significantly higher on the weekend. The other metrics (PM10, CO, NO2) show relative stability across all seven days of the week.

The higher weekend ozone is a known phenomenon often attributed to a combination of factors, including reduced NO emissions (which would normally destroy ozone overnight) and different atmospheric conditions.
